```python
#splitting the data and labeling the data
x=df.drop(['Delivery_Time_min'],axis=1)
y=df['Delivery_Time_min']
```

```python
print(x.columns)
#print(y.columns) ---considered as series sinec it is single column
```

```python
#splitting the data for train and text
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
#STEP 4: Model training
#1. lINEAR REGRESSION

from sklearn.linear_model import LinearRegression
#LinearRegression is a class, to use a class fruit we have to create object

lr=LinearRegression()
#Whatever the object name we give it is model(linear_regressor)

#trianing model
lr.fit(x_train, y_train)

#testing model(predicting results)
y_pred_lr=lr.predict(x_test)
print(y_pred_lr)  # these are results predicted by the model now we have to compare with actual values that is y_test

#model evaluation
#checking results based on y_test 

print("R^2 score of LR on training set:", lr.score(x_train, y_train))  #R^2 is not actually a acuracy but can we considered as acuracy and it cannot be 1 as it is considered as overfitting 
print("R^2 score of Lr on test set:", lr.score(x_test, y_test))

#checking for error (checking wrong prediction)
from sklearn.metrics import mean_absolute_error as mae,mean_squared_error as mse

print("---Linear Regression-Model evaluation--")
print("Mean Absolute Error(MAE):", mae(y_test, y_pred_lr))
print("Mean squared Error (MSE):", mse(y_test, y_pred_lr)) # to find minimum error we square it 
print("Root Mean squared Error(RMSE):", np.sqrt(mse(y_test, y_pred_lr))) #to make all the -ve values to +ve values
```

```python
#decision tree
#STEP 4: Model training
#1. Decision Tree REGRESSION

from sklearn.tree import DecisionTreeRegressor
#LinearRegression is a class, to use a class fruit we have to create object

dtr=DecisionTreeRegressor()
#Whatever the object name we give it is model(linear_regressor)

#trianing model
dtr.fit(x_train, y_train)

#testing model(predicting results)
y_pred_dtr=dtr.predict(x_test)
print(y_pred_dtr)  # these are results predicted by the model now we have to compare with actual values that is y_test

#model evaluation
#checking results based on y_test 

print("R^2 score of DT Ron training set:", dtr.score(x_train, y_train))  #R^2 is not actually a acuracy but can we considered as acuracy and it cannot be 1 as it is considered as overfitting 
print("R^2 score of DTR on test set:", dtr.score(x_test, y_test))

#checking for error (checking wrong prediction)
from sklearn.metrics import mean_absolute_error as mae,mean_squared_error as mse

print("---Linear Regression-Model evaluation--")
print("Mean Absolute Error(MAE):", mae(y_test, y_pred_dtr))
print("Mean squared Error (MSE):", mse(y_test, y_pred_dtr)) # to find minimum error we square it 
print("Root Mean squared Error(RMSE):", np.sqrt(mse(y_test, y_pred_dtr))) #to make all the -ve values to +ve values
```

```python
#4.support vector regressor

from sklearn.svm import SVR
svr=SVR()
svr.fit(x_train, y_train)
#testing model(predicting results)'
#predicts 30% of x_test for testing

y_pred_svr=svr.predict(x_test)
y_pred_svr
#model evaluation
#checking results based on y_test 

print("R^2 score of SVR on training set:", svr.score(x_train, y_train))  #R^2 is not actually a acuracy but can we considered as acuracy and it cannot be 1 as it is considered as overfitting 
print("R^2 score of SVR on test set:", svr.score(x_test, y_test))


#checking for error (checking wrong prediction)
from sklearn.metrics import mean_absolute_error as mae,mean_squared_error as mse

print("----SUPPORT VECTOR REGRESSION MODEL --")
print("Mean Absolute Error(MAE):", mae(y_test, y_pred_svr))
print("Mean squared Error (MSE):", mse(y_test, y_pred_svr)) # to find minimum error we square it 
print("Root Mean squared Error(RMSE):", np.sqrt(mse(y_test, y_pred_svr))) #to make all the -ve values to +ve values
```
